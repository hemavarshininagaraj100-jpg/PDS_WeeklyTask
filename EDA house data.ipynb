```python
if "date" in df.columns:
    df["date"] = pd.to_datetime(df["date"], errors="coerce")
    df["sale_year"] = df["date"].dt.year
    df["sale_month"] = df["date"].dt.month

    print("Date range:", df["date"].min(), "to", df["date"].max())
```

```python
if "price" in df.columns:
    fig, axes = plt.subplots(1, 2, figsize=(14, 5))

    sns.histplot(df["price"], bins=50, kde=True, ax=axes[0])
    axes[0].set_title("House Price Distribution")

    sns.boxplot(x=df["price"], ax=axes[1])
    axes[1].set_title("House Price Boxplot")

    plt.tight_layout()
    plt.show()

    print(df["price"].describe(percentiles=[.01, .05, .25, .5, .75, .95, .99]))
```

```python
numeric_cols = df.select_dtypes(include=np.number).columns.tolist()

df[numeric_cols].hist(figsize=(18, 14), bins=30, edgecolor="black")
plt.suptitle("Numeric Feature Distributions", y=1.02)
plt.tight_layout()
plt.show()
```

```python
numeric_df = df.select_dtypes(include=np.number)
corr = numeric_df.corr()

plt.figure(figsize=(14, 10))
sns.heatmap(corr, cmap="coolwarm", center=0, annot=False)
plt.title("Numeric Feature Correlation Matrix")
plt.tight_layout()
plt.show()

if "price" in corr.columns:
    display(corr["price"].sort_values(ascending=False).to_frame("correlation_with_price"))
```

```python
features = [
    "sqft_living", "bedrooms", "bathrooms", "floors",
    "condition", "grade", "waterfront", "yr_built"
]

features = [col for col in features if col in df.columns]

for col in features:
    plt.figure(figsize=(7, 4))
    sns.scatterplot(data=df, x=col, y="price", alpha=0.35)
    plt.title(f"Price vs {col}")
    plt.tight_layout()
    plt.show()
```

```python
features = [
    "sqft_living", "bedrooms", "bathrooms", "floors",
    "condition", "grade", "waterfront", "yr_built"
]

features = [col for col in features if col in df.columns]

for col in features:
    plt.figure(figsize=(7, 4))
    sns.scatterplot(data=df, x=col, y="price", alpha=0.35)
    plt.title(f"Price vs {col}")
    plt.tight_layout()
    plt.show()
```

```python
for col in ["bedrooms", "bathrooms", "floors", "condition", "grade", "waterfront"]:
    if col in df.columns:
        summary = (
            df.groupby(col, dropna=False)["price"]
              .agg(["count", "mean", "median"])
              .sort_values("median", ascending=False)
        )

        print(f"\nPrice summary by {col}:")
        display(summary)

        plt.figure(figsize=(8, 4))
        sns.boxplot(data=df, x=col, y="price")
        plt.title(f"Price by {col}")
        plt.tight_layout()
        plt.show()
```

```python
if "date" in df.columns and "price" in df.columns:
    monthly_price = (
        df.dropna(subset=["date"])
          .set_index("date")["price"]
          .resample("ME")
          .median()
    )

    monthly_price.plot(figsize=(12, 5), marker="o")
    plt.title("Monthly Median House Sale Price")
    plt.ylabel("Median price")
    plt.xlabel("Month")
    plt.tight_layout()
    plt.show()
```

```python
if {"lat", "long", "price"}.issubset(df.columns):
    plt.figure(figsize=(10, 7))

    points = plt.scatter(
        df["long"], df["lat"],
        c=df["price"], cmap="viridis",
        s=8, alpha=0.5
    )

    plt.colorbar(points, label="Price")
    plt.xlabel("Longitude")
    plt.ylabel("Latitude")
    plt.title("House Sale Prices by Location")
    plt.tight_layout()
    plt.show()
```
