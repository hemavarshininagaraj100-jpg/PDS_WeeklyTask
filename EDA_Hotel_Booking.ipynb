```python
print(type(plt))
```

```python
plt.figure(figsize=(9,5))

sns.lineplot(
    data=monthly_bookings,
    x="arrival_date_month",
    y="hotel",
    marker="o"
)

plt.title("Monthly Hotel Bookings")
plt.xlabel("Arrival Month")
plt.ylabel("Number of Bookings")
plt.tight_layout()

plt.show()
```
