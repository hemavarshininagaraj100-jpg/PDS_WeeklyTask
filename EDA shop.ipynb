```python
print(df.corr(numeric_only=True))
```

```python
sns.heatmap(df.corr(numeric_only=True), annot=True)
plt.show()
```

```python
plt.figure(figsize=(12,5))
plt.plot(df['date'], df['close'])
plt.title('Shopify Closing Price')
plt.xlabel('Date')
plt.ylabel('Closing Price')
plt.show()
```

```python
plt.scatter(df['open'], df['close'])
plt.title('Opening vs Closing Price')
plt.xlabel('Open Price')
plt.ylabel('Close Price')
plt.show()
```

```python
sns.histplot(df['volume'], bins=30)
plt.title('Trading Volume Distribution')
plt.show()
```

```python
sns.boxplot(data=df[['open','high','low','close']])
plt.title('Stock Price Distribution')
plt.show()
```

```python
df['daily_return'] = df['close'].pct_change() * 100

print(df[['date','close','daily_return']].head())
```

```python
plt.figure(figsize=(12,5))
plt.plot(df['date'], df['daily_return'])
plt.title('Daily Stock Returns')
plt.xlabel('Date')
plt.ylabel('Daily Return (%)')
plt.show()
```

```python
df['MA_20'] = df['close'].rolling(20).mean()
df['MA_50'] = df['close'].rolling(50).mean()

plt.figure(figsize=(12,5))
plt.plot(df['date'], df['close'], label='Close')
plt.plot(df['date'], df['MA_20'], label='20 Days')
plt.plot(df['date'], df['MA_50'], label='50 Days')

plt.legend()
plt.title('Shopify Moving Average')
plt.show()
```

```python
print("Highest Price:", df['high'].max())
print("Lowest Price:", df['low'].min())
print("Average Price:", df['close'].mean())
```

```python
print("Mean:", df['close'].mean())
print("Median:", df['close'].median())
print("Mode:", df['close'].mode()[0])
print("Standard Deviation:", df['close'].std())
print("Variance:", df['close'].var())
```
